# In 2025 who was the most valuable player on the Knicks? 
## and why was it Josh Hart?
### MVP will be determined by the follwing formula: 
### score = total rebounds + total assists + total steals + (field goals made * field goal percentage)



```python
query = """
SELECT 
players.full_name,
SUM(playerStats.REB) + SUM(playerStats.AST) + SUM(playerStats.STL) + SUM(playerStats.FGM * playerStats.FG_PCT) as score,
(SUM(playerStats.REB) + SUM(playerStats.AST) + SUM(playerStats.STL) + SUM(playerStats.FGM * playerStats.FG_PCT))/playerStats.min AS "score_per_min",
CAST(SUBSTR(playerStats.GAME_DATE_REAL, 1, 4) AS INTEGER) AS year,
SUM(playerStats.FGM) AS FGM,
SUM(playerStats.REB) AS Rebounds,
SUM(playerStats.AST) AS Assists,
AVG(playerStats.FG_PCT) AS "FG %"

FROM players
JOIN playerStats
ON players.id = playerStats.Player_ID

JOIN teams
ON teams.abbreviation = 
CASE
WHEN playerStats.MATCHUP LIKE '%@%' THEN
TRIM(SUBSTR(playerStats.MATCHUP, 1, INSTR(playerStats.MATCHUP, '@') - 1))
ELSE
TRIM(SUBSTR(playerStats.MATCHUP, 1, INSTR(playerStats.MATCHUP, 'vs.') - 1))
END

WHERE teams.full_name = 'New York Knicks' AND CAST(SUBSTR(playerStats.GAME_DATE_REAL, 1, 4) AS INTEGER) IN (2024, 2025)

GROUP BY players.id

ORDER BY score DESC
LIMIT 15;
"""

pd.read_sql_query(query, connection)
```

The query above selects statistics by using the SUM and AVG functions while also grouping together variables to make new variables. It pulls data from 3 previously defined tables. It intially pulls from the players dataset which only contains the player name. Then it joins playerStats on player_ID as well as teams on team abreviation being related to the abreviations in the matchup. It then filters for Knicks and the correct year and fianlly groups by playerID. Then it orders the data by the created score metric. Josh Hart, having the 2nd most score on the team also had one of the highest scores per minute, a signficantly higher score per minute than the number one "scorer" Karl Anthony Towns. It is easy to see the effort he puts in on the court pays its dividends and it is nice to see this point being argued for so strongly analytically. 

# Who is the most balenced offensive and defensive player in the NBA?

```python
query = """ 
SELECT 
players.full_name,
AVG(playerStats.plus_minus) as plus_minus,
SUM(playerStats.BLK) + SUM(playerStats.STL) + SUM(playerStats.REB) as "Defensive_Stats_total",
SUM(playerStats.pts) + SUM(playerStats.ast) as "Offensive_Stats_total",
SUM(playerStats.BLK) + SUM(playerStats.STL) + SUM(playerStats.REB) + SUM(playerStats.pts) + SUM(playerStats.ast) as "total_stats",
SUM(playerStats.fg_pct) as "FG %",
SUM(playerStats.FGM) as FGM,
SUM(playerStats.BLK) as Blocks,
SUM(playerStats.STL) as Steals,
sum(playerStats.min) as mins

from 
players

join playerStats
on players.id = playerStats.player_ID


group by playerStats.player_ID

HAVING mins > 1400

order by (total_stats) desc
LIMIT 15
"""

pd.read_sql_query(query, connection)
```

```python
query = """
SELECT AVG(total_min) AS mean_total_min
FROM (
    SELECT player_ID, SUM(min) AS total_min
    FROM playerStats
    GROUP BY player_ID
) AS player_totals
"""
pd.read_sql_query(query, connection)

```

The queries above selects many of the same variables as the query about Josh Hart and creates a few new ones to keep track of offensive, defensive and total stats. It pulls data from two tables, players and playerStats and joins on player_ID. I used a nested query to find the average sum of minutes for players in this dataset and then used that number to filter out players below that value to ensure players with significant playing time and therefore contribution, recieved the top ranks.
